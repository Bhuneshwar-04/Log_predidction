```python
from google_play_scraper import app, Sort , reviews_all
import plotly.express as px
import pandas as pd
import numpy as np
```

```python
nb_project = reviews_all('englishlearningspeaking.learnenglish.englishskillsbooster',
                         sleep_milliseconds=0,
                         lang='en',
                         country='us',
                         sort=Sort.MOST_RELEVANT )
```

```python
print(nb_project)
```

```python
df = pd.DataFrame(np.array(nb_project),columns=['review'])
df = df.join(pd.DataFrame(df.pop('review').tolist()))
df
```

```python
from transformers import pipeline
sentiment_pipeline = pipeline("sentiment-analysis", model="siebert/sentiment-roberta-large-english")
```

```python
print(sentiment_pipeline('I had tried more than 20 English learning apps like Duolingo, sivi ai learn, hello talk, memories speak but it didnt help me this is the only app that helped me a lot if your English is weak etc. If you cant speak English, you can try it, it will be a good app for you'))
```

```python
df['content'] = df['content'].astype(str)
df['content']
```

```python
df['result'] = df['content'].apply(lambda x: sentiment_pipeline(x))
```

```python
df['sentiments'] = df['result'].apply(lambda x: x[0]['label'])
df['sentiments']
```

```python
df['score'] = df['result'].apply(lambda x: x[0]['score'])
df['score']
```
